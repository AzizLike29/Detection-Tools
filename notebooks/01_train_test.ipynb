# Cell 3 - Test Kamera Laptop

```python
model = YOLO("yolov8n.pt")
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError("Laptop camera could not be opened")

print("Camera is running. Press Q in the video window to stop")

while True:
    ok, frame = cap.read()
    if not ok:
        break
    result = model.predict(frame, conf=0.5, verbose=False)[0]
    cv2.imshow("Test Camera YOLO", result.plot())
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

# Cell 4 - Train (jalankan hanya setelah foto sudah dilabel)

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
model.train(
    data=r"D:\DATA C\Projek\Detection Object - Yolo8\data.yaml",
    epochs=100,
    imgsz=640,
    batch=8,
    device="cpu",
)
```
